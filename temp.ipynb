```python
# load sample.gif

import sys
import os
import time
import numpy as np
import cv2

def load_gif(filename):
    cap = cv2.VideoCapture(filename)
    if not cap.isOpened():
        print("Error: could not open video.")
        sys.exit(1)

    frames = []
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frames.append(frame)
    cap.release()

    return frames
```

```python
gif = load_gif("sample.gif")
```
